## Figures

Figures are generated from the same equation outputs.

```python
figdir = ROOT / 'figures'
figdir.mkdir(exist_ok=True)
cash = pd.concat([eq11, eq12, eq13], ignore_index=True)
ax = cash.plot.bar(x='metric', y='percent', legend=False, color=['#355C7D', '#6C5B7B', '#C06C84'])
ax.set_ylabel('Percent')
ax.set_title('2019 U.S. reported-currency calibration')
plt.tight_layout(); plt.savefig(figdir / 'cash_calibration.png', dpi=160); plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for label, group in eq19.groupby('scenario_id'):
    axes[0].plot(group.n, group.loss_variance, marker='o', label=label)
axes[0].set(xlabel='Partitions N', ylabel='Loss variance', title='Equation 19')
axes[0].legend()
for label, group in eq20.groupby('scenario_id'):
    axes[1].plot(group.n, group.catastrophic_loss_probability, marker='o', label=label)
axes[1].set(xlabel='Partitions N', ylabel='Probability', title='Equation 20')
axes[1].legend()
plt.tight_layout(); plt.savefig(figdir / 'partition_risk.png', dpi=160); plt.show()
```

```python
ax = eq21.plot(x='n', y='objective', marker='o', legend=False)
best = eq21.loc[eq21.is_minimum].iloc[0]
ax.scatter([best.n], [best.objective], color='red', zorder=3, label='conditional minimum')
ax.set(xlabel='Partitions N', ylabel='Objective value', title='Equation 21 sensitivity scenario')
ax.legend(); plt.tight_layout(); plt.savefig(figdir / 'partition_objective.png', dpi=160); plt.show()
```

## Interpretation boundary

The numerical minimizer in Equation 21 is conditional on the supplied illustrative probabilities, risk weight, feasible values of $N$, and cost function. It is not an empirical recommendation.
